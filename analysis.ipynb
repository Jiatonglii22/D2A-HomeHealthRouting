# Home Health Nurse Routing Optimization 

## Objective 
The goal of this project is to assign patients to nurses and determine daily nurse routes while considering: 

- nurse availability 
- patient appointment dates 
- visit duration 
- travel time 
- continuity of care 

This notebook will serve as a sensitivity analysis to see how the results of the optimization change as we value the continuity of care penalty higher or lower. 

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 

from Optimization_Model.model import build_model 
```

```python
DATA_DIR = './Datasets/Forecast/' # or './Datasets/Dummy/' (regenerate travel_car.csv to match) 

patients_df = pd.read_csv(DATA_DIR + 'patient_appointments.csv')
nurses_df = pd.read_csv(DATA_DIR + 'nurse_availability.csv')
distance_df = pd.read_csv('./Datasets/Processed/travel_car.csv')
```

# Model Overview 
### Decision Variables

$ x_{ijd} = 1 $ if nurse $i$ is assigned to patient $j$ on day $d$, 0 otherwise.

$ z_{ij} = 1 $ if nurse $i$ has ever been matched to patient $j$, 0 otherwise.

### Objective

Minimize travel cost and continuity of care penalty:

$$
\min
\sum_{ijd}
c_{ij}x_{ijd}
+
w_c * \sum_j(\sum_i z_{ij} - 1) 
$$

where $w_c$ is the weight of the continuity of care penalty 

```python
# sensitivity analysis 
w_c_values = [0, 0.5, 1, 5, 10, 50]

for w_c in w_c_values: 
    # model 
    model, x, z = build_model(nurses_df, patients_df, distance_df, w_c) 

    model.optimize()

    # verify optimal solution is found 
    if model.status == 2: 
        print("Optimal solution found")
        print(f"Objective: {model.ObjVal:.2f}")
```
